```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

# LOAD DATA

```python
var = sns.load_dataset("tips")
var
```

# COUNT PLOT

```python
sns.countplot(x="sex", data=var, hue="smoker", palette="bwr", saturation=0.5)
plt.show()
```

# VIOLIN PLOT

```python
sns.violinplot(x="day", y="total_bill", data=var, hue="time", linewidth=2, palette="Dark2", saturation=0.2, color="r")
plt.show()
```

```python
sns.violinplot(x="time", y="total_bill", data=var, order=["Dinner","Lunch"], inner="quart")
plt.show()
```

```python
sns.violinplot(x="day", y="total_bill", data=var, hue="sex", split=True, scale="width")
plt.show()
```

```python
sns.violinplot(x="total_bill", y="day", data=var, hue="sex")
plt.show()
```

```python
sns.violinplot(x=var["total_bill"])
plt.show()
```

# PAIR PLOT

```python
sns.pairplot(var)
plt.show()
```

```python
sns.pairplot(var, vars=["total_bill","tip"], hue="sex", hue_order=["Female","Male"], palette="BuGn", kind="kde", diag_kind="hist")
plt.show()
```

# STRIP PLOT

```python
sns.stripplot(x="day", y="total_bill", data=var, hue="sex", palette="rocket_r", linewidth=1, 
              edgecolor="r", jitter=1, size=14, marker="<", alpha=0.4)
plt.show()
```

```python
sns.stripplot(y=var["total_bill"])
plt.show()
```

# BOX PLOT

```python
sns.set(style="whitegrid")
sns.boxplot(x="day", y="total_bill", data=var, hue="sex", color="g", order=["Fri", "Sun", "Thur", "Sat"], 
            showmeans=True, meanprops={"marker":"+", "markeredgecolor":"b"}, linewidth=3, orient="v")
plt.show()
```

```python
sns.set(style="whitegrid")
sns.boxplot(x=var["total_bill"])
plt.show()
```

# CAT PLOT

```python
sns.catplot(x="tip", y="size", data=var, hue="sex", palette="BuPu", kind="point")
plt.show()
```

# STYLING PLOT

```python
sns.set_style("white")
sns.set_context("poster", font_scale=1)
sns.set_palette("cool")
# plt.figure(figsize=(12,3))
sns.barplot(x="day", y="total_bill", data=var)
sns.despine()
plt.show()
```

# MULTIPLE(Facet Grid) PLOT

```python
fg= sns.FacetGrid(var, col="day", hue="sex", palette="summer")
fg.map(plt.bar, "total_bill", "tip", edgecolor="g").add_legend()
plt.show()
```
