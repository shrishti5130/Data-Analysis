```python
# 5. Array Manipulation
import numpy as np

# Original array
arr = np.array([1, 2, 3, 4, 5, 6])

print("Original Array:")
print(arr)


# 1. reshape()
reshape_arr = arr.reshape(2, 3)
print("\n1. reshape():")
print(reshape_arr)


# 2. flatten()
flatten_arr = reshape_arr.flatten()
print("\n2. flatten():")
print(flatten_arr)


# 3. ravel()
ravel_arr = reshape_arr.ravel()
print("\n3. ravel():")
print(ravel_arr)


# 4. transpose()
transpose_arr = reshape_arr.T
print("\n4. transpose():")
print(transpose_arr)


# Two arrays for joining
a = np.array([[1, 2],
              [3, 4]])

b = np.array([[5, 6],
              [7, 8]])


# 5. concatenate()
concat_arr = np.concatenate((a, b), axis=0)
print("\n5. concatenate():")
print(concat_arr)


# 6. vstack()
vstack_arr = np.vstack((a, b))
print("\n6. vstack():")
print(vstack_arr)


# 7. hstack()
hstack_arr = np.hstack((a, b))
print("\n7. hstack():")
print(hstack_arr)
```

# NumPy Random

```python
# 1. Random number between 0 and 1
print(np.random.rand())

# 2. Multiple random numbers
print(np.random.rand(5))

# 3. Random integers
print(np.random.randint(1, 10))

# 4. For multiple numbers:
print(np.random.randint(1, 10, 5))

# 5. Random numbers from normal distribution
np.random.randn(5)
```

# DATA TYPE

# shape vs reshape()

```python
# 1. shape --> shape tells you the current structure of the array.
a = np.array([[1, 2, 3],[4, 5, 6]])
print(a.shape)

# 2. reshape() --> reshape() changes the structure/shape of the array.
a = np.array([1, 2, 3, 4, 5, 6])
b = a.reshape(2, 3)
print(b)
```

```python
# 1D array
a = np.array([1, 2, 3, 4, 5, 6])
print("1D:", a)

# 1D → 2D
b = a.reshape(2, 3)
print("2D:", b)

# 2D → 1D
c = b.reshape(-1)
print("1D again:", c)
```

# Indexing AND Slicing

```python
# Indexing
# 1-D
a = np.array([10, 20, 30, 40, 50])
print(a[0])   # 10
print(a[2])   # 30
print(a[-1])  # 50
# 2-D
a = np.array([[10, 20, 30],
              [40, 50, 60]])
print(a[0, 1])
```

```python
# slicing
# 1-D
a = np.array([10, 20, 30, 40, 50])
print(a[1:4])
print(a[:3])    # First 3 elements
print(a[2:])    # From index 2 to end
print(a[::2])   # Every 2nd element
print(a[::-1])  # Reverse array
# 2-D
a = np.array([[10, 20, 30],
              [40, 50, 60]])
print(a[0:2, 1:3])
```

# ARRAY Function

```python
# 1. np.where() — Search --> np.where() is used to find the positions (indexes) where a condition is true.
a = np.array([10, 20, 30, 40, 50])
result = np.where(a > 30)
print(result)
```

```python
# 2. np.sort() — Sort --> np.sort() sorts the elements in ascending order by default.
a = np.array([40, 10, 30, 20])
print(np.sort(a))

# Descending order --> NumPy does not have a direct descending=True parameter. You can reverse the sorted array:
print(np.sort(a)[::-1])
```

```python
# 3. np.searchsorted() — Search in a Sorted Array --> searchsorted() finds the index where a new value should be inserted while keeping the array sorted.
a = np.array([10, 20, 30, 40, 50])
print(np.searchsorted(a, 35))

# Multiple values
a = np.array([10, 20, 30, 40, 50])
print(np.searchsorted(a, [15, 35, 45]))
```

```python
# 4. Filter --> Filtering means selecting only the elements that satisfy a condition.
# Boolean filtering
a = np.array([10, 20, 30, 40, 50])
result = a[a > 25]
print(result)
```

# Insert & delete

```python
# 1. np.insert() — Insert --> np.insert() adds an element at a specific index.
# Syntax --> np.insert(array, index, value)
a = np.array([10, 20, 30, 40])
b = np.insert(a, 2, 25)
print(b)
# np.insert() returns a new array. It does not directly change the original array.
```

```python
a = np.array([[10, 20],
              [30, 40]])
b = np.insert(a, 1, [50, 60], axis=0) 
# Here, [50, 60] is inserted at row index 1.
print(b)
```

```python
a = np.array([[10, 20],
              [30, 40]])
b = np.insert(a, 1, [50, 60], axis=1)
# Here, [50, 60] is inserted at column index 1.
print(b)
```

```python
# 2. np.append() — Append --> np.append() adds elements at the end of an array.

a = np.array([10, 20, 30])
b = np.append(a, 40)
print(b)
```

```python
a = np.array([[10, 20],
              [30, 40]])
b = np.append(a, [[50, 60]], axis=0)
print(b)
```

```python
a = np.array([[10, 20],
              [30, 40]])
b = np.append(a, [[50],[60]], axis=1)
print(b)
```

```python
# 3. np.delete() — Delete --> np.delete() removes an element from a specific index.
# Syntax --> np.delete(array, index)
a = np.array([10, 20, 30, 40])
b = np.delete(a, 2)
print(b)
```

```python
a = np.array([[10, 20],
              [30, 40]])
b = np.delete(a, 0, axis=0)
print(b)
```

```python
a = np.array([[10, 20],
              [30, 40]])
b = np.delete(a, 1, axis=1)
print(b)
```
