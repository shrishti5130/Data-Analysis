# MERGE

```python
df1 = pd.DataFrame({"id": [1, 2, 3],"name": ["A", "B", "C"]})
df2 = pd.DataFrame({"id": [1, 2, 3],"salary": [20000, 30000, 40000]})
pd.merge(df1, df2, on="id")
```

```python
pd.merge(df1, df2, on="id", how="outer", indicator=True)
```

```python
pd.merge(df1, df2, left_index=True, right_index=True, suffixes=("name", "python"))
```

# CONCAT

```python
sr1 = pd.Series([1,2,3,4])
sr2 = pd.Series([11,22,33,44])
pd.concat([sr1,sr2])
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2], axis=1, join="inner")
```

```python
pd.concat([df1, df2], keys=["DataFrame1", "DataFrame2"])
```

# GROUPBY

```python
var = pd.DataFrame({"name":["a","b","c","d","a","b","a","b","c","c"],
                   "sub1":[12,13,14,12,13,14,15,23,25,16],
                   "sub2":[25,26,27,28,29,30,25,34,35,40]})
var
```

```python
gro = var.groupby("name")
gro
```

```python
for x,y in gro:
    print(x)
    print(y)
```

```python
gro.get_group("a")
```

```python
gro.min()
```

```python
gro.max()
```

```python
gro.mean()
```

```python
li = list(gro)
li
```

# JOIN

```python
var1 = pd.DataFrame({"A":[12,13,14,11], "B":[15,16,17,18]}, index=[1, 2, 3,4])
var2 = pd.DataFrame({"C":[22,23,24,22], "D":[25,26,27,28]}, index=[1, 2, 3,4])
var1.join(var2)
```

```python
var1.join(var2, how="left")
```

```python
var1.join(var2, lsuffix="_left", rsuffix="_right")
```

# APPEND

```python
pd.concat([var1, var2], ignore_index=True)
```

# MELT()

```python
df = pd.DataFrame({"days": [1,2,3,4,5,6],"Math": [80, 90,36,86,74,85],"English": [70, 85,67,57,45,57]})
df
```

```python
pd.melt(df)
```

```python
pd.melt(df, id_vars=["days"], var_name="python", value_name="Sh")
```

# PIVOT TABLE

```python
df = pd.DataFrame({"Name": ["A", "A", "B", "B"],"Subject": ["Math", "English", "Math", "English"],"Marks": [80, 70, 90, 85]})
df
```

```python
pd.pivot_table(df,values="Marks",index="Name",columns="Subject",aggfunc="mean")
```
