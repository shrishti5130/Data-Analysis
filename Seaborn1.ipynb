# LINE PLOT

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

var1 = [1,2,3,4,5,6,7]
var2 = [2,3,4,1,5,6,7]

var3 = pd.DataFrame({"var1":var1, "var2":var2})

sns.lineplot(x="var1", y="var2", data=var3)

plt.show()
```

# LOAD DATA

```python
var = sns.load_dataset("penguins")
var
```

# LINE PLOT

```python
sns.lineplot(x="bill_length_mm", y="flipper_length_mm", data=var, hue="sex", palette="Accent", markers=["o",">"])
plt.grid()
plt.title("Python")
plt.show()
```

# Histogram Plot

```python
sns.displot(var["bill_length_mm"], bins=[34,36,38,40,42,44,46],kde=True, rug=True, color="red")
plt.show()
```

# BAR PLOT

```python
sns.barplot(x="island", y="bill_length_mm", data=var, hue="sex")
plt.show()
```

```python
sns.set(style="dark")
orders=["Biscoe", "Dream", "Torgersen"]
sns.barplot(x="island", y="bill_length_mm", data=var)
plt.show()
```

```python
sns.barplot(x="bill_depth_mm", y="bill_length_mm", data=var, orient="h")
plt.show()
```

# SCATTER PLOT

```python
m={"Male":"*", "Female":"o"}
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm", data=var, hue="sex", style="sex", size="sex", sizes=(20,40),
               palette="gist_ncar", alpha=0.5, markers=m)
plt.show()
```

# HEAT MAP

```python
# data using array
a = np.linspace(1,10,20).reshape(4,5)
a
```

```python
f={"fontsize":10, "color":"r"}
v = sns.heatmap(a, vmin=0, vmax=10, cmap="gist_heat", annot=True, annot_kws=f, linewidth=6, 
                linecolor="black", cbar=True, xticklabels=False, yticklabels=False)
v.set(xlabel="Python", ylabel="Cube")
sns.set(font_scale=4)
plt.show()
```
