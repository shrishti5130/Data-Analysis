# SERIES

```python
import pandas as pd
x = [3,4,5,6,7,8]
var = pd.Series(x, index=["a","b","c","d","e","f"], dtype="float", name="Python")
print(var)
print(type(var))
print(var[3])
```

```python
dic = {"name":["Python", "Java"], "Rank":[1,4]}
var1 = pd.Series(dic)
print(var1)
```

```python
s1 = pd.Series(12, index=[1,2,3,4,5,6])
s2 = pd.Series(12, index=[1,2,3,4])
print(s1+s2)
```

# DATAFRAME

```python
d={"a":[1,2,3,4], "b":[5,6,7,8]}
d1 = pd.DataFrame(d, columns=["a"], index=["a","b","c","d"])
print(type(d1))
print(d1)
```

```python
import pandas as pd

s1 = pd.Series(["A", "B", "C"])
s2 = pd.Series([20, 21, 22])
s3 = pd.Series([85, 90, 78])

df = pd.DataFrame({"Name": s1,"Age": s2,"Marks": s3})

print(df)
```

```python
sr={"a":pd.Series(["A", "B", "C"]), "b":pd.Series([20, 21, 22])}
df = pd.DataFrame(sr)
print(df)

```

# ARITHMETIC OPERATIONS

```python
var = pd.DataFrame({"a":[1,2,3,4], "b":[5,6,7,8]})
var
var["c"] = var["a"]+var["b"]
print(var)
```

# INSERT

```python
var = pd.DataFrame({"A":[1,2,3,4], "B":[5,6,7,8]})
var
```

```python
var.insert(1, "C", var["A"])
var
```

```python
var.insert(1, "D", ["A","B","C","D"])
var
```

# WRITE THEN CREATE CSV FILE

```python
dis = {"a":[1,2,3,4], "b":[5,6,7,8], "c":[4,6,3,7]}
d = pd.DataFrame(dis)
print(d)
d.to_csv("file_1.csv")
```

```python
dis = {"a":[1,2,3,4], "b":[5,6,7,8], "c":[4,6,3,7]}
d = pd.DataFrame(dis)
print(d)
d.to_csv("file_2.csv",index=False)
```

```python
dis = {"a":[1,2,3,4], "b":[5,6,7,8], "c":[4,6,3,7]}
d = pd.DataFrame(dis)
print(d)
d.to_csv("file_3.csv", index=False, header=["sh","at","ak"])
```
