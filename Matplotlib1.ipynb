```python
import matplotlib.pyplot as plt
import numpy as np
```

# BAR PLOT

```python
x = ["Python", "Jave", "C", "C++"]
y = [76,43,75,85]
plt.xlabel("Languages")
plt.ylabel("Number")
plt.title("BAR PLOT", fontsize = 20)
c = ["b","y","r","g"]
plt.bar(x,y, width=0.5, color=c, align="edge", edgecolor="r", linewidth=5, linestyle=":", alpha=0.5, label="Shrishti")
plt.legend()
plt.show()
```

```python
# MULTIPLE BAR in one PLOT

x = ["Python", "Jave", "C", "C++"]
y = [76,43,75,85]
z = [56,32,54,24]

width=0.2
p = np.arange(len(x)) # LIST TO ARRAY
p1 = [j+width for j in p]

plt.xlabel("Languages")
plt.ylabel("Number")
plt.title("BAR PLOT", fontsize = 20)

plt.bar(p,y, width, color="g", label="Shrishti")
plt.bar(p1,z, width, color="r", label="Shrishti")

plt.xticks(p+width/2, x, rotation=20) # control the labels and positions on the X-axis
plt.legend()
plt.show()
```

```python
# HORIZONTAL BAR PLOT
x = ["Python", "Jave", "C", "C++"]
y = [76,43,75,85]
plt.xlabel("Number")
plt.ylabel("Languages")
plt.title("BAR PLOT", fontsize = 20)
plt.barh(x,y)
plt.legend()
plt.show()
```

# SCATTER PLOT

```python
x =[4,5,6,3,7,3]
y =[5,7,8,3,6,7]

colors=["y","g","r","k","b","r"]
sizes =[234,213,432,543,232,234]

plt.xlabel("Days")
plt.ylabel("Number")
plt.title("SCATTER PLOT", fontsize = 20)

plt.scatter(x,y, c=colors, s=sizes, alpha=0.5, marker="*", edgecolor="g", linewidth=2)
plt.show()
```

```python
x =[4,5,6,3,7,3]
y =[5,7,8,3,6,7]

colors=[12,43,23,21,22,32]
sizes =[234,213,432,543,232,234]

plt.xlabel("Days")
plt.ylabel("Number")
plt.title("SCATTER PLOT", fontsize = 20)

plt.scatter(x,y, c=colors, s=sizes, cmap="viridis")
t =plt.colorbar()
t.set_label("COLORBAR", fontsize=15)
plt.show()
```

```python
x =[4,5,6,3,7,3]
y =[5,7,8,3,6,7]
y1 =[5,6,4,6,4,6]

colors=[12,43,23,21,22,32]
sizes =[234,213,432,543,232,234]

plt.xlabel("Days")
plt.ylabel("Number")
plt.title("SCATTER PLOT", fontsize = 20)

# for multiple scatter plot in one graph
plt.scatter(x,y, c=colors, s=sizes, cmap="viridis")
plt.scatter(x,y1, c=colors, s=sizes, cmap="viridis")

t =plt.colorbar()
t.set_label("COLORBAR", fontsize=15)
plt.show()
```

# PIE CHART

```python
x=[10,20,30,40]
y=["C++", "C", "Java", "Python"]
ex=[0.0,0.2,0.0,0.0]
c=["r","b","g","y"]
plt.pie(x, labels=y, explode=ex, colors=c, autopct="%1.1f%%", shadow=True, radius=0.9, 
        labeldistance=1.2, startangle=180, textprops={"fontsize":15}, counterclock=False,
       wedgeprops={"linewidth":4, "edgecolor":"k"}, center=(1,3), rotatelabels=False)
plt.title("PIE CHART")
plt.legend(loc=3)
plt.show()
```

```python
# Point chart
plt.pie([1])
plt.show()
```

```python
# Donut piechart or multiple piechart
x=[10,20,30,40]
x1=[40,30,20,10]
y=["C++", "C", "Java", "Python"]
c=["r","b","g","y"]

plt.pie(x, labels=y, radius=1.5)
plt.pie(x1, labels=y, radius=1, colors=c)

plt.show()
```

```python
# ring piechart
x=[10,20,30,40]
x1=[40,30,20,10]
y=["C++", "C", "Java", "Python"]
c=["r","b","g","y"]

plt.pie(x, labels=y, radius=1.5)
plt.pie([1], colors="w")

plt.show()
```

```python
# another way to ring piechart
x=[10,20,30,40]
x1=[40,30,20,10]
y=["C++", "C", "Java", "Python"]
c=["r","b","g","y"]

plt.pie(x, labels=y, radius=1.5)
cr = plt.Circle(xy=(0,0), radius=1, facecolor="w")
plt.gca().add_artist(cr)

plt.show()
```

# STEM PLOT

```python
x=[2,3,1,5,4,7]
y=[7,5,8,4,3,2]
plt.stem(x,y, linefmt=":", markerfmt="ro", bottom=0, basefmt="k", 
         label="Python", orientation="horizontal")
plt.legend()
plt.show()
```

# BOX PLOT

```python
x=[10,20,30,40,50,60,70,120]
plt.boxplot(x, notch=False, vert=True, widths=0.5, labels=["Python"], patch_artist=True,
           showmeans=True, sym="g+", whis=1, boxprops=dict(color="r"), 
           capprops=dict(color="r"), whiskerprops=dict(color="r"),
           flierprops=dict(markeredgecolor="y"))
plt.show()
```

```python
# multiple box plot
x=[10,20,30,40,50,60,70,120]
x1=[20,30,40,50,60,70,80]
y=[x,x1]
plt.boxplot(y, notch=False, vert=True, widths=0.5, labels=["Python", "C"], patch_artist=True,
           showmeans=True, sym="g+", whis=1, boxprops=dict(color="r"), 
           capprops=dict(color="r"), whiskerprops=dict(color="r"),
           flierprops=dict(markeredgecolor="y"))
plt.show()
```

# AREA and STACK PLOT

```python
x=[1,2,3,4,5,]
area1=[1,2,4,4,2]
area2=[4,3,5,2,1]
area3=[4,2,5,2,3]
l=["area1","area2","area3"]
plt.stackplot(x,area1, area2, area3, labels=l, colors=["r","g","y"], baseline="zero")
plt.title("Stack Plot", fontsize=20)
plt.xlabel("a-axis")
plt.ylabel("y-axis")
plt.legend(loc=2)
plt.show()
```

# STEP PLOT

```python
x=[1,2,3,4,5]
y=[1,2,3,4,5]
plt.step(x,y, marker="o", color="r", ms= 10, mfc="g", label="Python")
plt.title("STEP PLOT")
plt.xlabel=("x-axis")
plt.ylabel=("y-axis")
plt.legend()
plt.grid()
plt.show()
```

# FILL_BETWEEN PLOT

```python
import numpy as np
x = np.array([1, 2, 3, 4, 5])
y = np.array([1, 2, 3, 4, 5])
plt.plot(x, y, color="r")
plt.fill_between(x, y,where=(x >= 2) & (x <= 4), alpha=0.5)
plt.title("Linear Fill_BETWEEN")
plt.show()
```
